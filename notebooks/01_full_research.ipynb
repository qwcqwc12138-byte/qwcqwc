# Stock Market Analysis and Portfolio Optimization

This notebook is a thin, reproducible interface over the tested package code. Run the data-fetch command first, then execute these cells from the repository root.

```python
from pathlib import Path

from market_portfolio.analysis import (
    adjusted_price_matrix,
    asset_performance_summary,
    load_market_csv,
)
from market_portfolio.config import load_config
from market_portfolio.research import run_full_research

DATA = Path("../data/processed/yahoo/prices_2023-01-01_2026-08-14.csv")
CONFIG = Path("../config/universe.yaml")
OUTPUT = Path("../artifacts")
```

```python
frame = load_market_csv(DATA)
prices = adjusted_price_matrix(frame)
asset_performance_summary(prices, risk_free_rate=0.038).style.format(
    "{:.2%}", subset=["total_return", "annual_return", "annual_volatility", "max_drawdown"]
)
```

```python
result = run_full_research(DATA, load_config(CONFIG), OUTPUT)
{method: backtest.metrics for method, backtest in result.backtests.items()}
```
